```

from numpy import *

def kahn(A):
	d = len(A) # A is a numpy array
	indegree = [ sum(A[i]) for i in range(d) ]
	inputs = [ j for j in range(d) if indegree[j] == 0 ]
	order = []
	while inputs:
		j = inputs.pop(0)
		order.append(j)
		for i in range(d):
			if A[i,j]:
				indegree[i] -= 1
				if indegree[i] == 0: inputs.append(i)
	if len(A) == len(order):
		Atop = zeros(shape(A), dtype = int)
		for i in range(d):
			for j in range(d): Atop[i,j] = A[order[i],order[j]]
		return Atop
	else: print('Graph is not a DAG')
	
A = zeros((8,8), dtype = int)
A[6,4] = A[7,2] = 1
A[4,5] = A[4,3] = A[2,5] = A[2,3] = 1
A[5,1] = A[3,1] = A[5,0] = A[3,0] = 1

kahn(A)
```
