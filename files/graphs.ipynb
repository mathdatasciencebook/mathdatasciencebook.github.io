```
 
from numpy import *

A = zeros((8,8), dtype = int)
A[4,5] = A[4,3] = A[2,5] = A[2,3] = A[6,4] = 1
A[5,1] = A[3,1] = A[5,0] = A[3,0] = A[7,2] = 1
A
```

```

def kahn(A):
	iso = empty(0, dtype = int)
	indegree = sum(A, axis = 1)
	inputs = where(indegree == 0)[0]
	while len(inputs):
		iso = append(iso, inputs)
		for j in inputs: indegree[ A[:,j] != 0 ] -= 1
		not_iso = full(len(A), True)
		not_iso[iso] = False
		inputs = where((indegree == 0) & not_iso)[0]
	if len(A) == len(iso): 
		return A[ meshgrid(iso, iso, indexing = 'ij') ]
	else: print('Graph is not a DAG.')
```
