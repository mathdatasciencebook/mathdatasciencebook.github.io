```

# projection of column vector v
# onto column space of U
# with orthonormal columns
	
def project_to_ortho(U, v):
	x = dot(U.T, v)      # reduced
	return dot(U, x)     # projected
```

```

def gs(A):
	d, N = shape(A)
	U = zeros((d,0))
	# step through columns
	for v in A.T:
		Pv = project_to_ortho(U, v)
		if not allclose(v, Pv):
			u = (v-Pv) / sqrt(dot(v-Pv, v-Pv))
			U = column_stack([U, u])
	return U
```
