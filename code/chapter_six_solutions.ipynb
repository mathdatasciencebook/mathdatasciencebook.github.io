```

##########################
# Confidence Interval - proportions
##########################

from scipy.stats import norm as Z

def z_score(alpha): return Z.ppf(1 - alpha)

def confidence_interval(sample, alpha):
	phat = mean(sample)
	n = len(sample)
	z = z_score(alpha/2)
	disc = sqrt(z**2 + 4*n*phat*(1-phat))
	L = (2*n*phat + z**2 - z*disc) / (2*(n+z**2))
	U = (2*n*phat + z**2 + z*disc) / (2*(n+z**2))
	return L, U
```
