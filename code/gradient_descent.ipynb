```

from numpy import *

def newton(loss, grad, curv, w, num_iter):
	g = grad(w)
	c = curv(w)
	for _ in range(num_iter):
		w -= g/c
		try: trajectory = array([*trajectory, [w,loss(w)]])
		except NameError: trajectory = array([[w,loss(w)]])
		g = grad(w)
		c = curv(w)
		if allclose(g,0): break
	return trajectory
```

```

def loss(w): return w**4 - 6*w**2 + 2*w     # f(w)
def grad(w): return 4*w**3 - 12*w + 2       # f'(w)
def curv(w): return 12*w**2 - 12            # f''(w)

u0 = -2.72204813
w0 = 2.45269774
num_iter = 20
trajectory = newton(loss, grad, curv, w0, num_iter)
```

```

from matplotlib.pyplot import *

def plot_descent(a, b, loss, curv, delta, trajectory):
	w = arange(a, b, delta)
	plot(w, loss(w), c = 'red', lw = 1)
	plot(w, curv(w), '--', c = 'blue', lw = 1)
	plot(*trajectory.T, c = 'green', lw = 1)
	scatter(*trajectory.T, s = 10)
	title(f'num_iter = {len(trajectory)}')
	grid()
	show()
```

```

ylim(-15,10)
delta = .01
plot_descent(u0, w0, loss, curv, delta, trajectory)
```

```

def gd(loss, grad, w, lr, num_iter):
	g = grad(w)
	for _ in range(num_iter):
		w -= lr * g
		try: trajectory = array([*trajectory, [w,loss(w)]])
		except NameError: trajectory = array([[w,loss(w)]])
		g = grad(w)
		if allclose(g,0): break
	return trajectory
```

```

a = -2.72204813
w0 = 2.45269774
L = 76.914552
lr = 1/L
num_iter = 100
trajectory = gd(loss, grad, w0, lr, num_iter)

ylim(-15,10)
delta = .01
plot_descent(a, w0, loss, curv, delta, trajectory)
```
