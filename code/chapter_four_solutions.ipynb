```

from numpy import *
from matplotlib.pyplot import *

for a, b in [(-pi/4, pi/4), (-pi/2, pi/2)]:
	theta = linspace(a, b, 500)
	m, L = min(-cos(theta)), max(-cos(theta))   # min and max of f''
	plot(theta, cos(theta), label = 'cos')
	plot(theta, 1 + m/2*theta**2, label = 'lower')
	plot(theta, 1 + L/2*theta**2, label = 'upper')
	legend()
	grid()
	show()
```

```

from numpy import *
from matplotlib.pyplot import *
from scipy.stats import entropy
from scipy.optimize import newton

def H(p): return entropy([p,1-p])

p = arange(0.001, 1, .001)
plot(p, H(p), label = 'H(p)')
plot(p, sqrt(p*(1-p)), label = 'sqrt(p(1-p))')
legend()
grid()
show()

def diff(p): return H(p) - sqrt(p*(1-p))
r1 = newton(diff, 0.1)
r2 = newton(diff, 0.9)
r1, r2
```

```

from numpy import *
from matplotlib.pyplot import *
from scipy.stats import entropy

q = 0.7
def I(p,q): return entropy([p,1-p],[q,1-q])

p = arange(0.001, 1, .001)
plot(p, I(p,q), label = 'I(p,q)')
plot(p, 2*(p-q)**2, label = '2(p-q)^2')
legend()
grid()
show()
```

```

from numpy import *

def f(x): return sin(x)
def g(r): return 1/(1+exp(-r))
def h(s): return s**2
def df(x): return cos(x)
def dg(r): return g(r)*(1-g(r))
def dh(s): return 2*s

d = 4
w = full((d,d), None)
w[1,0] = w[2,1] = w[3,2] = 1
activate = [None, f, g, h]

x = forprop([pi/4], w)     # second version, as in the text
print(x)

g_mat = full((d,d), None)
g_mat[1,0] = lambda t: df(t)
g_mat[2,1] = lambda t: dg(t)
g_mat[3,2] = lambda t: dh(t)

delta = backprop(x, [1.0], g_mat)
print(delta)
```
