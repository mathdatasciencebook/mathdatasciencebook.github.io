```

def row_stack(arrs):
	m = max(ndim(x) for x in arrs) - 1
	rows = []
	for x in arrs:
		if ndim(x) == m:
			rows.append(x)      # x is itself a single row
		else:
			rows += list(x)     # x is a stack
	return array(rows)
```

```

from numpy import *
from numpy.linalg import pinv

u = array([1,2,3,4,5])
v = array([6,7,8,9,10])
w = array([11,12,13,14,15])
A = column_stack([u,v,w])
Aminus = pinv(A)

150*Aminus  # matches the stated A^-

p = array([-72,11,94,177,260])
q = array([4,3,2,1,0])
3*dot(Aminus,p), 3*dot(Aminus,q)          # matches x^- for p, q
allclose(dot(A,dot(Aminus,p)), p)         # True: Ax=p is solvable
allclose(dot(A,dot(Aminus,q)), q)         # True: Ax=q is solvable

b = array([-9,-3,3,9,10])
xm = dot(Aminus,b)
15*xm                                     # matches stated x^-
dot(A,xm)                                 # (-8,-3,2,7,12), not equal to b

B = column_stack([p,q])
Bminus = pinv(B)
1300*Bminus                               # matches stated B^-
52*dot(Bminus,u), 26*dot(Bminus,v), 52*dot(Bminus,w)   # match stated solutions

C = A.T
f = array([-100,-3,94])
g = array([2,1,0])
Cminus = pinv(C)
150*Cminus                                # matches stated C^-
50*dot(Cminus,f), 50*dot(Cminus,g)        # match stated solutions

D = column_stack([f,g])
Dminus = pinv(D)
564*Dminus                                # matches stated D^-
rows = [array([1,6,11]), array([2,7,12]), array([3,8,13]), array([4,9,14]), array([5,10,15])]
[dot(Dminus,r) for r in rows]             # match stated solutions (94,47,94,47,94 denominators)
```
