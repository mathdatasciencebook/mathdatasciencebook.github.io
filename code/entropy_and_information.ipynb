```
 
%matplotlib ipympl
from numpy import *
from  matplotlib.pyplot import *

def I(p,q): return entropy([p,1-p], [q,1-q])

p = arange(0, 1, .01)
q = arange(0, 1, .01)
p, q = meshgrid(p, q)

ax = axes(projection = '3d')
ax.plot_surface(p, q, I(p,q), cmap = 'cool')

corners = array([[0,0], [1,0], [1,1], [0,1], [0,0]])
ax.plot(*corners.T, lw = .5, c = 'k')

ax.axis('off')
show()
```
