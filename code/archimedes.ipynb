```python
from sympy import *
from IPython.display import display, Math

def sqroot(P):
    x,y = P
    return ((1+x)/sqrt(2+2*x),y/sqrt(2+2*x))

J = (0,1)
s = 'no'

def theta(P, n):
    x,y = P
    if n == 1: return sqrt(2-2*x)
    else: return simplify(2*theta(sqroot(P),n-1))
        
for n in range(1,7): 
    a = 2*theta(J,n)
    if 223/71 < N(a) < 22/7: s = 'yes'
    display(Math(rf'{s}: 2\theta_{n} = {latex(a, order= 'none')}'))
    
```

```python
theta2, theta2star = theta(J,2)
display(2*theta2,2*theta2star)
```
