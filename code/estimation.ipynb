```

from numpy import *
set_printoptions(legacy = '1.25')
from numpy.random import default_rng as rng

def Angle(x,y):
	a = dot(x,y)
	b = dot(x,x)
	c = dot(y,y)
	theta = arccos(a / sqrt(b*c))
	return degrees(theta)

samples = rng().normal

mu, sdev, d = 0, 1, 784

for _ in range(20):
	x = samples(mu,sdev,d)
	y = samples(mu,sdev,d)
	print(Angle(x,y))
```

```

samples = rng().binomial

n, p, d = 1, .5, 784 # one coin toss
# n, p, d = 3, .5, 784 # three coin tosses

for _ in range(20):
	x = samples(n,p,d)
	y = samples(n,p,d)
	print(Angle(x,y))
```
