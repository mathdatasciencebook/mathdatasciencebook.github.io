```python
from numpy import *

from numpy.random import default_rng as rng
samples = rng().random

#dxN matrix

d = 3
N = 5

f = samples((d,))
g = samples((d,))
p = samples((N,))
q = samples((N,))

A = outer(f,p) + outer(g,q)
A
```

```python

Q = dot(A.T, A)
tr = Q.trace()
tr
```

```python
from numpy.linalg import eigh

lamda, U = eigh(Q)
lamda[-1] * lamda[-2]
```

```python
Qprime = Q - tr* eye(N)/2
Qprime
```

```python
from numpy.linalg import norm, eigh

u = 10*p - sqrt(2)*q

a = norm(dot(Qprime, u))
b = norm(u)

delta = a/b

(tr/2)**2 - delta**2
```
