```

mu = array([9, 4.5])
w = array([13.3488, 8.25])  # direction of best fit
n = sqrt(dot(w,w))

t = arange(-8, 9, .1)
line = mu + outer(t, w)/n

scatter(*dataset.T, s = 5)
scatter(*mu, c = 'red')
plot(*line.T, c = 'green')
grid()
show()
```

```

mu = mean(dataset, axis = 0)
Q = cov(dataset.T, bias = True)
a, b, c = Q[0,0], Q[0,1], Q[1,1]
lam_plus = (a+c)/2 + sqrt(((a-c)/2)**2 + b**2)
w = array([lam_plus - c, b])

t = arange(-3, 4, .1)
line = mu + outer(t, w)/norm(w)

scatter(*dataset.T, s = 5)
scatter(*mu, c = 'red')
plot(*line.T, c = 'green')
grid()
show()
```

```

mu = mean(dataset, axis = 0)
Q = cov(dataset.T, bias = True)
Sigma_inv = diag(1/sqrt(diag(Q)))
standardized = (dataset - mu) @ Sigma_inv.T

Qprime = corrcoef(dataset.T)  # equals [[1, 1/sqrt(2)], [1/sqrt(2), 1]]

scatter(*standardized.T, s = 5)
grid()
show()
```
