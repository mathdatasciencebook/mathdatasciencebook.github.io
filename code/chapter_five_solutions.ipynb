```

from scipy.integrate import quad
from scipy.special import factorial

n, k = 8, 5
coeff = factorial(n+1)/(factorial(k)*factorial(n-k))
def posterior(p): return  coeff*p**k*(1-p)**(n-k)
quad(posterior, 0.5, 0.7)
```

```

from sympy import *

a, b, c = symbols('a, b, c', positive=True)

# eq:uncor3: X,Y uncorrelated
eq1 = a + 2*b + c - 1
eq2 = a - b - (a-c)*(a+b)
solve([eq1,eq2], [a,b])

# eq:uncor4: Z=XY, Y uncorrelated
eq3 = (a-b) - (a-b)*(a+b)
solve([eq1,eq3], [a,b])
```

```

from scipy.stats import norm as Z

def z_score(alpha): return Z.ppf(1 - alpha)

# Z = (X - mu)/sigma

a = z_score(1 - 0.3)   #  0.3 = P(X<1) = P(Z<a)
b = z_score(1 - 0.4)   #  0.4 = P(X<2) = P(Z<b)

# a = (1-mu)/sigma
# b = (2-mu)/sigma
sigma = (2-1)/(b-a)

mu = 1 - a*sigma
mu, sigma**2
```

```

from sklearn import datasets
from numpy import *
from scipy.stats import multivariate_normal as Z

iris = datasets.load_iris()
dataset, labels = iris['data'], iris['target']

# class means and variances (Exercise prob:irismQ)
mus, Qs = [], []
for c in [0,1,2]:
	sub = dataset[labels == c]
	mus.append(mean(sub, axis = 0))
	Qs.append(cov(sub.T, bias = True))

def classify_probs(x):
	dens = array([ Z.pdf(x, mean = mus[k], cov = Qs[k]) for k in range(3) ])
	return dens/sum(dens)      # equal priors cancel top and bottom (eq:bayes2)

preds = array([ argmax(classify_probs(x)) for x in dataset ])
sum(preds == labels)
```
